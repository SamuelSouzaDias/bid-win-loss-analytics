Criando um Catalogo


```python
df_clientes = spark.read.format("excel")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("dataAddress", "Planilha1!A1:L1545")\
    .load("/Volumes/raw/bid/bids/clientes.xlsx")
```

```python
# Extrair nomes das colunas da primeira linha
header_row = df_clientes.first()
column_names = [header_row[i] for i in range(len(header_row))]

# Renomear colunas e remover a linha de cabeçalho
df_clientes_clean = df_clientes.toDF(*column_names)
df_clientes_clean = df_clientes_clean.filter(df_clientes_clean.cd_id_anonimo != 'cd_id_anonimo')
```

```python
df_clientes_clean\
    .write\
    .mode("overwrite")\
    .option("mergeSchema", True)\
    .format("delta")\
    .saveAsTable("bronze.bid.clientes")
```
