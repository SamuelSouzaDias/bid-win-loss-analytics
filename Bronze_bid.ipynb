Criando um Catalogo


```python
df_excel = spark.read.format("excel")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .option("dataAddress", "Bronze!A1:K1583")\
    .load("/Volumes/raw/bid/bids/bronze.bid.xlsx")
```

```python
# Extrair nomes das colunas da primeira linha
header_row = df_excel.first()
column_names = [header_row[i] for i in range(len(header_row))]

# Renomear colunas e remover a linha de cabeçalho
df_excel_clean = df_excel.toDF(*column_names)
df_excel_clean = df_excel_clean.filter(df_excel_clean.idbid != 'idbid')
```

```python
df_excel_clean\
    .write\
    .mode("overwrite")\
    .option("mergeSchema", True)\
    .format("delta")\
    .saveAsTable("bronze.bid.bids")
```
